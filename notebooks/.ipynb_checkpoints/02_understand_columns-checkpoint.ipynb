```python
import os
from pathlib import Path

ROOT = Path.cwd()
while not (ROOT / "pyproject.toml").exists():
    ROOT = ROOT.parent
os.chdir(ROOT)
```

```python
import pandas as pd

from iran_shock.columns import CODE_COLUMNS, EVENT_COLUMNS

df = pd.read_csv(
    ROOT / "data/raw/20260228120000.export.CSV",
    sep="\t",
    header=None,
    names=EVENT_COLUMNS,
    dtype=dict.fromkeys(CODE_COLUMNS, str),
)
df.shape
```

```python
df.head()
```

```python
# Top countries by where the event happened
df["ActionGeo_CountryCode"].value_counts().head(10)
```

```python
# Overall average tone across the whole file
df["AvgTone"].astype(float).mean()
```

```python
# Iran-only average tone
df.loc[df["ActionGeo_CountryCode"] == "IR", "AvgTone"].astype(float).mean()
```

```python
# How coarse is AvgTone? Look at its distribution — one number per event, no nuance.
df["AvgTone"].astype(float).describe()
```

```python
# Actor1CountryCode is CAMEO (3-letter): IRN, USA, ISR
df["Actor1CountryCode"].value_counts().head(5)
```

```python
# QuadClass — 1 verbal coop, 2 material coop, 3 verbal conflict, 4 material conflict.
df["QuadClass"].value_counts()
```

```python
df["EventRootCode"].value_counts().head(10)
```

```python
df["ActionGeo_Type"].value_counts()
```

```python
# Volume of coverage ≈ how big the news is. Use this to weight events, not just count them.
df["NumMentions"].astype(int).describe()
df.nlargest(5, df["NumMentions"].astype(int).name)[["SOURCEURL", "NumMentions", "AvgTone"]]
```

```python
df["GoldsteinScale"].astype(float).describe()
# -10..+10 — strikes are very negative, talks positive. Pre-baked from event type, not learned.
```

```python
df["QuadClass"].value_counts()  # values are 3, 4, ... not "3", "4"
# Read the URLs — these should be strike/missile/casualty headlines, not noise.
```

```python
me = df[
    df["ActionGeo_CountryCode"].isin(
        ["IR", "IS", "US", "SA", "AE", "QA", "IZ", "KU", "BA", "LE", "SY", "YM"]
    )
    & df["QuadClass"].isin([3, 4])  # ints, not strings
]
len(me), me["AvgTone"].astype(float).mean()
me["SOURCEURL"].head(10).tolist()
```
