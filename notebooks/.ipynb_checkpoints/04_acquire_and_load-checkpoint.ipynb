## Ingestion — what `iran-shock-ingest-gdelt` produces

The script walks GDELT's deterministic 15-minute URL pattern (`gdeltv2/YYYYMMDDHHMMSS.export.CSV.zip`), downloads each file with retries (exponential backoff on transient errors), and writes to `data/raw/gdelt/date=YYYY-MM-DD/{stamp}.export.CSV`. The atomic write pattern (`.part` temp file + `os.replace`) means `dest` only exists at full size — safe to interrupt and re-run.

Per-day metrics land in `data/exports/ingest_metrics.csv` (committed). One row per partition:

````python
metrics = pd.read_csv("data/exports/ingest_metrics.csv")
print(f"days covered: {len(metrics)}")
print(f"total files:  {metrics['file_count'].sum():,}")
print(f"total events: {metrics['event_count'].sum():,}")
metrics.head()
````

## Load — how `iran-shock-build` materialises the gdelt table

Build runs:

```sql
CREATE OR REPLACE TABLE gdelt AS
SELECT * FROM read_csv('data/raw/gdelt/**/*.export.CSV',
    delim='\t', header=false, names=[...61 cols], all_varchar=true);
```

**Two design choices worth internalising:**

- **`all_varchar=true`.** Every column comes in as text. Casts happen later in the views (`avgtone::DOUBLE`, `strptime(day,'%Y%m%d')::DATE`). A single malformed value in one row can't poison the column's inferred type.
- **Hive partition column for free.** DuckDB auto-detects the `date=YYYY-MM-DD/` folders and adds a typed `date DATE` column — the table ends up with **62 columns**, not 61. This mirrors `PARTITIONED BY (date string)` in Athena (Stage 3): same shape locally and in the cloud.

````python
schema = con.execute("DESCRIBE gdelt").df()
print(f"columns: {len(schema)}")
print("types:", schema["column_type"].value_counts().to_dict())
schema.tail(5)   # last row should be the partition column 'date DATE'
````

````python
con.execute("SELECT count(*) AS events FROM gdelt").df()
````

Sample rows — note `Day` is the event date (mostly current, a small share backfilled from earlier dates); `DATEADDED` is when GDELT first detected it.

````python
con.execute("""
SELECT day, dateadded, actiongeo_countrycode, quadclass, avgtone,
       substr(sourceurl, 1, 60) AS url_excerpt
FROM gdelt LIMIT 5
""").df()
````

## Partition coverage

The Hive `date` column makes day-by-day filters cheap (`WHERE date BETWEEN ...`). Sanity-check that the window has no gaps:

````python
con.execute("""
SELECT min(date) AS first_partition,
       max(date) AS last_partition,
       count(DISTINCT date) AS distinct_days,
       date_diff('day', min(date), max(date)) + 1 AS span_days
FROM gdelt
""").df()
````

If `distinct_days == span_days`, every day in the window has files. Any gap is worth investigating — usually a GDELT-side outage that re-running the ingest CLI will fill once GDELT republishes.
