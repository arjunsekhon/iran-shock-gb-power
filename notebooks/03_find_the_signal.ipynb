```python
import pandas as pd

from iran_shock.columns import CODE_COLUMNS, EVENT_COLUMNS

df = pd.read_csv(
    ROOT / "data/raw/20260228120000.export.CSV",
    sep="\t",
    header=None,
    names=EVENT_COLUMNS,
    dtype=dict.fromkeys(CODE_COLUMNS, str),
)
df.shape
```

```python
from iran_shock.config import MIDDLE_EAST

print(df["QuadClass"].dtype, df["QuadClass"].value_counts().to_dict())
print(df["ActionGeo_CountryCode"].dtype)
print(df["ActionGeo_CountryCode"].value_counts().head(10).to_dict())
print("MIDDLE_EAST =", MIDDLE_EAST)
print("ME matches:", df["ActionGeo_CountryCode"].isin(MIDDLE_EAST).sum())
print("Quad matches:", df["QuadClass"].isin([3, 4]).sum())
```
